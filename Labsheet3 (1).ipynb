```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create sample student data
data = {
    "Student ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Student Name": ["Aman", "Riya", "Rahul", "Priya",
                     "Karan", "Neha", "Arjun", "Simran"],
    "Gender": ["Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female"],
    "Department": ["BCA", "BCA", "BTech", "BCA",
                   "BTech", "BCA", "BTech", "BTech"],
    "Attendance": [85, 92, 70, 95, 65, 88, 78, 90],
    "Study Hours": [5, 7, 3, 8, 2, 6, 4, 7],
    "Assignment Marks": [80, 90, 60, 95, 55, 85, 70, 88],
    "Internal Assessment Marks": [78, 92, 58, 94, 50, 82, 68, 90],
    "Python Marks": [85, 95, 62, 96, 48, 88, 72, 91],
    "DBMS Marks": [80, 92, 65, 94, 45, 84, 70, 89],
    "AI Marks": [82, 94, 60, 96, 50, 86, 75, 92],
    "Final Exam Marks": [84, 93, 61, 95, 47, 85, 70, 90]
}

df = pd.DataFrame(data)

# 2. Display the complete data
print("STUDENT DATA")
print(df)

# 3. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 4. Calculate Average Marks
df["Average Marks"] = df[
    ["Python Marks", "DBMS Marks", "AI Marks", "Final Exam Marks"]
].mean(axis=1)

# 5. Identify high-performing students
print("\nHIGH-PERFORMING STUDENTS")
high = df[df["Average Marks"] >= 85]
print(high[["Student Name", "Average Marks"]])

# 6. Identify low-performing students
print("\nLOW-PERFORMING STUDENTS")
low = df[df["Average Marks"] < 60]
print(low[["Student Name", "Average Marks"]])

# 7. Subject-wise performance
print("\nSUBJECT-WISE AVERAGE MARKS")
subjects = [
    "Python Marks",
    "DBMS Marks",
    "AI Marks",
    "Final Exam Marks"
]

subject_average = df[subjects].mean()
print(subject_average)

# 8. Department-wise performance
print("\nDEPARTMENT-WISE AVERAGE MARKS")
department_average = df.groupby("Department")["Average Marks"].mean()
print(department_average)

# 9. Attendance patterns
print("\nATTENDANCE INFORMATION")
print(df[["Student Name", "Attendance"]])

print("\nAVERAGE ATTENDANCE")
print(df["Attendance"].mean())

# 10. Study Hours and Marks
print("\nSTUDY HOURS AND AVERAGE MARKS")
print(df[["Student Name", "Study Hours", "Average Marks"]])

# 11. Attendance and Performance
print("\nATTENDANCE AND PERFORMANCE")
print(df[["Student Name", "Attendance", "Average Marks"]])

# ---------------- VISUALIZATIONS ----------------

# 12. Subject-wise performance
plt.bar(subject_average.index, subject_average.values)
plt.title("Subject-wise Average Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=20)
plt.show()

# 13. Department-wise performance
plt.bar(department_average.index, department_average.values)
plt.title("Department-wise Average Performance")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()

# 14. Attendance pattern
plt.bar(df["Student Name"], df["Attendance"])
plt.title("Student Attendance")
plt.xlabel("Student")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=45)
plt.show()

# 15. Study Hours vs Average Marks
plt.scatter(df["Study Hours"], df["Average Marks"])
plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")
plt.show()

# 16. Attendance vs Performance
plt.scatter(df["Attendance"], df["Average Marks"])
plt.title("Attendance vs Performance")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.show()
```
