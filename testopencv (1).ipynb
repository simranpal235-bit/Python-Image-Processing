```python
import cv2
img = cv2.imread(r"C:\Users\princ\OneDrive\Pictures\Screenshots\Screenshot 2026-10-05 212819.png")
print(img.shape)
import matplotlib.pyplot as plt
img_rgb= cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("on")
plt.show()
```

```python
import cv2
cap= cv2.VideoCapture(0)
while(True):
    ret,frame=cap.read()
    cv2.imshow('frame',frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
      break
cap.release()
cv2.destroyAllWindows()
```
