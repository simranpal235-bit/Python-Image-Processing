```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\princ\OneDrive\Pictures\Screenshots\Screenshot 2026-09-14 233559.png")

print(img.shape)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```
